```python
# here is the question; type over this to ask something new

system_prmpt = """" 
You are Social Media Marketing Consultant. Explain which of the links in the website make sense to use in a social media post about the company to make users
aware about the company's potential. Also put a snarky line at the end of the explanation.

Then generate the social media post content about the company (in a professional tone).

"""

def user_prompt_getting(url):
    links = fetch_website_links(url)   # or contents = fetch_website_contents(url)
    return f"""
Website URL: {url}

Here are links found on the page:
{links}

Your task: Explain which of the links in the website make sense to use in a social media post about the company to make users
aware about the company's potential.

Explain like This :-
    
[
    ["link":"www.website.com", "Mentionable:" "Yes" , "Reason":"Main website link"],
    \n
    ["link":"www.website/blog", "Mentionable:" "No" , "Reason":"Not main website link"]
]

Remember, no Maybe in the mentionablility of links, if can't say Yes definitely, say No. Only pages links to 
main website, social media and careers are to be considered for a Yes.
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
def analysis(url):

    stream = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "system", "content": system_prmpt},
            {"role": "user", "content": user_prompt_getting(url)}
        ],
        stream=True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

analysis("https://www.cursor.com/")
```
